###  Kategorik değişkenleri category yap

```python
# 1. Kategorik değişkenleri category yap
categoric_cols = ['PERSON_HOME_OWNERSHIP', 'LOAN_INTENT', 'LOAN_GRADE', 'CB_PERSON_DEFAULT_ON_FILE']

for col in categoric_cols:
    df[col] = df[col].astype("category")

# 2. Object görünümlü ama aslında sayısal olanları float'a çevir
num_but_car = ['LOAN_INT_RATE', 'LOAN_PERCENT_INCOME']

for col in num_but_car:
    df[col] = (
        df[col].astype(str)
               .str.replace(',', '.', regex=False)
               .str.replace('%', '', regex=False)
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 3. Kontrol: Hedef değişken int olarak kalsın
df['LOAN_STATUS'] = df['LOAN_STATUS'].astype(int)

# 4. Son durum dtype kontrolü
print(df.dtypes)
```
