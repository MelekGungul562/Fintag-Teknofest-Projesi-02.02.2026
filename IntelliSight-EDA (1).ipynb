```python
#############################
# AYKIRI DEĞER ANALİZİ
#############################

def outlier_thresholds(dataframe, col_name, q1=0.05, q3=0.95):
    quartile1 = dataframe[col_name].quantile(q1)
    quartile3 = dataframe[col_name].quantile(q3)
    interquantile_range = quartile3 - quartile1
    up_limit = quartile3 + 1.5 * interquantile_range
    low_limit = quartile1 - 1.5 * interquantile_range
    return low_limit, up_limit


def check_outlier(dataframe, col_name, q1=0.05, q3=0.95):
    low_limit, up_limit = outlier_thresholds(dataframe, col_name, q1, q3)
    return ((dataframe[col_name] < low_limit) | (dataframe[col_name] > up_limit)).any()



def replace_with_thresholds(dataframe, variable, q1=0.05, q3=0.95):
    low_limit, up_limit = outlier_thresholds(dataframe, variable, q1, q3)
    dataframe.loc[dataframe[variable] < low_limit, variable] = low_limit
    dataframe.loc[dataframe[variable] > up_limit, variable] = up_limit
```

```python
def outlier_summary(dataframe, num_cols, q1=0.05, q3=0.95):
    summary = []

    for col in num_cols:
        if check_outlier(dataframe, col):
            low, up = outlier_thresholds(dataframe, col, q1, q3)
            mask = (dataframe[col] < low) | (dataframe[col] > up)
            outliers = dataframe.loc[mask, col]

            summary.append({
                "Column": col,
                "Outlier_Count": outliers.shape[0],
                "Outlier_Min": outliers.min(),
                "Outlier_Max": outliers.max(),
                "Lower_Limit": low,
                "Upper_Limit": up
            })

    return pd.DataFrame(summary)

# Kullanım
outlier_table = outlier_summary(df, num_cols)
display(outlier_table)
```

```python
def outlier_thresholds(dataframe, col_name, q1=0.05, q3=0.95):
    quartile1 = dataframe[col_name].quantile(q1)
    quartile3 = dataframe[col_name].quantile(q3)
    interquantile_range = quartile3 - quartile1
    up_limit = quartile3 + 1.5 * interquantile_range
    low_limit = quartile1 - 1.5 * interquantile_range
    return low_limit, up_limit


def check_outlier(dataframe, col_name, q1=0.05, q3=0.95):
    low_limit, up_limit = outlier_thresholds(dataframe, col_name, q1, q3)
    return ((dataframe[col_name] < low_limit) | (dataframe[col_name] > up_limit)).any()



def replace_with_thresholds(dataframe, variable, q1=0.05, q3=0.95):
    low_limit, up_limit = outlier_thresholds(dataframe, variable, q1, q3)
    dataframe.loc[dataframe[variable] < low_limit, variable] = low_limit
    dataframe.loc[dataframe[variable] > up_limit, variable] = up_limit
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_recall_fscore_support, accuracy_score, confusion_matrix

# 1) Model olasılıkları (RandomForest örneği)
proba = rf.predict_proba(X_test)[:, 1]   # class=1 (kredi verildi) olasılığı

# 2) Belirli eşiklerde metrikleri hesaplayan yardımcı fonksiyon
def metrics_at_threshold(p, y_true, thr):
    y_hat = (p >= thr).astype(int)
    acc = accuracy_score(y_true, y_hat)
    prec, rec, f1, _ = precision_recall_fscore_support(y_true, y_hat, average="binary", zero_division=0)
    tn, fp, fn, tp = confusion_matrix(y_true, y_hat).ravel()
    return {"threshold": thr, "accuracy": acc, "precision": prec, "recall": rec, "f1": f1,
            "TP": tp, "FP": fp, "FN": fn, "TN": tn}

# 3) 0.20–0.80 aralığında 0.02 adımla tarayalım (istersen aralığı değiştir)
thresholds = np.round(np.arange(0.20, 0.81, 0.02), 2)
rows = [metrics_at_threshold(proba, y_test, t) for t in thresholds]
thr_table = pd.DataFrame(rows).sort_values("f1", ascending=False).reset_index(drop=True)

print(thr_table.head(20))   # en iyi 10 sonucu gör
```

```python
best_idx = thr_table["f1"].idxmax()
best_thr = float(thr_table.loc[best_idx, "threshold"])

print(f"Best F1 threshold: {best_thr:.2f}")
print(thr_table.loc[best_idx, ["precision","recall","f1","accuracy","TP","FP","FN","TN"]])

# Bu eşikte detaylı rapor
y_pred_best = (proba >= best_thr).astype(int)
print("Confusion Matrix @best thr:\n", confusion_matrix(y_test, y_pred_best))
```

```python
target_recall = 0.78
candidates = thr_table[thr_table["recall"] >= target_recall]
if len(candidates):
    pick = candidates.sort_values("precision", ascending=False).iloc[0]
    thr_recall_pref = float(pick["threshold"])
    print(f"Chosen threshold for recall≥{target_recall}: {thr_recall_pref:.2f}")
    print(pick[["precision","recall","f1","accuracy","TP","FP","FN","TN"]])
else:
    print("No threshold meets the target recall; relax the target or check the range.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_fscore_support, accuracy_score, confusion_matrix

# 1) Test olasılıkları (class=1)
proba = rf.predict_proba(X_test)[:, 1]

# 2) Eşik aralığı
thresholds = np.linspace(0.20, 0.80, 61)  # 0.20 -> 0.80 arası, 0.01 adım

rows = []
for t in thresholds:
    y_pred = (proba >= t).astype(int)
    acc = accuracy_score(y_test, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(
        y_test, y_pred, average="binary", zero_division=0
    )
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
    rows.append([t, acc, prec, rec, f1, tp, fp, fn, tn])

thr_df = pd.DataFrame(rows, columns=["thr","accuracy","precision","recall","f1","TP","FP","FN","TN"])

# 3) Hedef: müşteri kazanımı → recall yüksek, ama precision da çok düşmesin
# ör: recall ≥ 0.76 koşuluyla precision en yüksek eşiği seç
target_recall = 0.76
cands = thr_df[thr_df["recall"] >= target_recall]
pick = cands.sort_values("precision", ascending=False).iloc[0] if len(cands) else thr_df.loc[thr_df["f1"].idxmax()]
best_thr = float(pick["thr"])
print("Chosen threshold:", round(best_thr, 2))
print(pick[["precision","recall","f1","accuracy","TP","FP","FN","TN"]])

# 4) Grafik: threshold vs precision/recall/F1
plt.figure(figsize=(8,5))
plt.plot(thr_df["thr"], thr_df["precision"], label="Precision")
plt.plot(thr_df["thr"], thr_df["recall"],    label="Recall")
plt.plot(thr_df["thr"], thr_df["f1"],        label="F1")
plt.axvline(best_thr, ls="--", color="gray", label=f"chosen thr={best_thr:.2f}")
plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Threshold Tuning (class=1)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# 5) Seçilen eşikte karışıklık matrisi görmek istersen:
y_pred_best = (proba >= best_thr).astype(int)
print("Confusion matrix @chosen thr:\n", confusion_matrix(y_test, y_pred_best))
```
